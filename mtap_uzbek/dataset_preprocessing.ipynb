```python
import pandas as pd
df = pd.read_csv('kunuz.csv')
len(df)
```

```python
df = df[['title', 'author', 'category', 'date', 'tags', 'body', 'url']]
```

```python
count_v = df['author'].str.endswith('в', na=False).sum()
count_va = df['author'].str.endswith('ва', na=False).sum()
count_other = df['author'].str.contains('Kun.uz Admin')
print("Ends with 'в':", count_v)
print("Ends with 'ва':", count_va)
print(f'other {len(count_other)}')
```

```python
# Step 1: Create a new column for gender (if it doesn't exist yet)
if 'gender' not in df.columns:
    df['gender'] = 'unknown'

# Step 2: Apply your rules ONLY to the 'gender' column
df.loc[df['author'] == 'Kun.uz Admin', 'gender'] = 'unknown'
df.loc[df['author'] == '—', 'gender'] = 'unknown'
df.loc[df['author'] == 'Ўроқбоев Шаҳзод', 'gender'] = 'male'
df.loc[df['author'] == 'Жамила Фадл', 'gender'] = 'female'

# General rules based on Uzbek name endings
df.loc[df['author'].str.endswith('ва', na=False), 'gender'] = 'female'   # e.g. Назокатхон Эшонова
df.loc[df['author'].str.endswith('в', na=False), 'gender'] = 'male'      # e.g. Отабек Раҳимов
# (you can add more endings like 'ой', 'бек', etc. if needed)
```

```python
df = df[['title', 'gender', 'category', 'date', 'tags', 'body', 'url']]
```

```python
df.sample(10, random_state=6)
```

```python
df['date'] = pd.to_datetime(df['date'].str.split(' / ').str[-1], 
                            format='%d.%m.%Y', 
                            errors='coerce').dt.strftime('%d.%m.%Y')

print(df['date'].eq('—').sum())
df.loc[df['date'] == '—'] = 'unknown'
print(df['date'].eq('—').sum())

df = df[df['category'] != 'Kun.uz']

df['category'] = df['category'].replace({
    'Жамият': 'Jamiyat',
    'Ўзбекистон': 'O‘zbekiston',
    'Жаҳон': 'Jahon'
})

df = df[
    (df['body'] != 'NaN') &
    (df['category'] != '—') &
    (df['title'] != '—')
]

import ast
df['tags'] = df['tags'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)
print(df['new_tags'].dtype)

df[df['body'] == 'NaN'].head()
```

```python
def extract_three(tags):
    if not isinstance(tags, list):
        return [None, None, None]

    t = tags[:3]
    t = [tag.lstrip('#') if isinstance(tag, str) else None for tag in t]

    while len(t) < 3:
        t.append(None)

    return t

df[['tag1', 'tag2', 'tag3']] = pd.DataFrame(
    df['tags'].apply(extract_three).to_list()
)
```

```python
df.to_csv('clean_kunuz.csv', index=False)
```
